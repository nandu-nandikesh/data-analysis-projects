# **Capping the outliers**

```python
def cap_outliers_iqr(dataframe):
    df_capped = dataframe.copy()

    for col in df_capped.columns:
        Q1 = df_capped[col].quantile(0.25)
        Q3 = df_capped[col].quantile(0.75)
        IQR = Q3 - Q1

        ''' threshold values '''
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        ''' capping the values outside thresholds '''
        df_capped[col] = np.where(df_capped[col] < lower_bound, lower_bound,
                                  np.where(df_capped[col] > upper_bound, upper_bound, df_capped[col]))

    return df_capped


X = df.drop('fetal_health', axis=1)
y = df['fetal_health']

''' Apply capping to features '''
X_capped = cap_outliers_iqr(X)

df_capped = pd.concat([X_capped, y], axis=1)

print("Original shape:", df.shape)
print("After capping outliers:", df_capped.shape)
```

# **Feature Engineering**

```python

''' Creating 3 new feature ratio of light,severe and prolonged decelerations by contractions '''

df['ratio_light_to_uterine_contractions'] = df['light_decelerations'] / (df['uterine_contractions'] + 1e-5)
df['ratio_severe_to_uterine_contractions'] = df['severe_decelerations'] / (df['uterine_contractions'] + 1e-5)
df['ratio_prolonged_to_uterine_contractions'] = df['prolongued_decelerations'] / (df['uterine_contractions'] + 1e-5)

''' Calculating Total decelerations '''

df['total_decelerations'] = df['light_decelerations'] + df['severe_decelerations'] + df['prolongued_decelerations']

''' Creating 2 new Normalized feature variability'''

df['normalized_short_term_variability'] = df['mean_value_of_short_term_variability'] / (df['baseline value'] + 1e-5)
df['normalized_long_term_variability'] = df['mean_value_of_long_term_variability'] / (df['baseline value'] + 1e-5)

'''Creating a new feature Interaction to find out healthy, reactive fetus '''

df['acceleration_fetal_movement_interaction'] = df['accelerations'] * df['fetal_movement']

''' This gives the range, wide range or narrow range from
maximum value of the fetal heart rate histogram and subtracts the minimum value '''

df['histogram_range'] = df['histogram_max'] - df['histogram_min']

''' Compares how many peaks(modes) exist in the histogram to how many zero-value(bins with no count) bins exist. '''

df['histogram_peak_to_zero_ratio'] = df['histogram_number_of_peaks'] / (df['histogram_number_of_zeroes'] + 1e-5)
df
```
